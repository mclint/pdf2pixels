# pdf2pixels quickstart

Render the included two-page PDF and display each result inline. The notebook works from either the repository root or the `examples` directory.

```python
from pathlib import Path

from IPython.display import Image, display

from pdf2pixels import convert

sample_pdf = Path("sample.pdf")
if not sample_pdf.exists():
    sample_pdf = Path("examples/sample.pdf")
sample_pdf
```

## Render every page

Omitting `pages` renders all pages in document order.

```python
pages = list(convert(sample_pdf))
[(page.page, page.format, page.width, page.height) for page in pages]
```

```python
for page in pages:
    display(Image(data=page.data))
```

## Select and reorder pages

Public page numbers are 1-based, and explicit order is preserved.

```python
selected = list(convert(sample_pdf, pages=[2, 1], dpi=72))
[(page.page, page.width, page.height) for page in selected]
```

## Convert PDF bytes to JPEG

Path and in-memory byte inputs use the same streaming API.

```python
jpeg_page = next(
    convert(sample_pdf.read_bytes(), pages=[1], format="jpeg", dpi=100)
)
display(Image(data=jpeg_page.data))
```
